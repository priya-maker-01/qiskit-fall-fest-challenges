# Qiskit Fall Fest 2026 — Challenge 2

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import math

simulator = AerSimulator()

# Challenge 2: Biased quantum coin
qc_1 = QuantumCircuit(1, 1)

qc_1.ry(math.pi / 3, 0)
qc_1.measure(0, 0)

display(qc_1.draw("mpl"))

print("Prediction: approximately 75% zero and 25% one")
```

```python
job_1 = simulator.run(
    transpile(qc_1, simulator),
    shots=100
)

counts_1 = job_1.result().get_counts()

print("Actual results:", counts_1)

plot_histogram(counts_1)
plt.show()
```

```python
# Second biased coin: 25% zero and 75% one
qc_2 = QuantumCircuit(1, 1)

qc_2.ry(2 * math.pi / 3, 0)
qc_2.measure(0, 0)

display(qc_2.draw("mpl"))
```

```python
job_2 = simulator.run(
    transpile(qc_2, simulator),
    shots=100
)

counts_2 = job_2.result().get_counts()

print("Prediction: approximately 25% zero and 75% one")
print("Actual results:", counts_2)

plot_histogram(counts_2)
plt.show()
```

```python
print("CIRCUIT COMPARISON")
print("Circuit 1 (75% zero, 25% one):", counts_1)
print("Circuit 2 (25% zero, 75% one):", counts_2)
```
